```python
numbers = [0, 10, 20, 30, 40, 50, 60, 70]
uni_name = "North Carolina State University"

print(numbers[1:8:3])
print(numbers[6:2])
print(numbers[-3::-3])
print(uni_name[15:21])
# Write a single line of code to create a tuple named tup1 that contains the elements 20, 40 and 60 (in that order) by slicing the list provided in class.
tup1 = tuple(numbers[2:7:2]) # starts at index 2, goes up to but not including index 7, and takes every second element (step of 2). 
print(tup1)

#Write a loop to count and output the number of lowercase letter "o"s contained in the string provided in class.
total = 0
for char in uni_name:
    if char == 'o':
        total += 1
print(total)

total = [char for char in uni_name if char == 'o']
print(len(total))
```
