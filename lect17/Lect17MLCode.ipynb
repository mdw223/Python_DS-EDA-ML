```python
#Import necessary packages
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
tips = pd.read_csv('tips.csv')
tips.head()
```

```python
import seaborn as sns
sns.regplot(x='total_bill', y = 'tip', ci = None, data = tips).set(title = 'Relationship between Tip and Total Bill')
```

```python
from sklearn import linear_model
#create linear regression object
lr = linear_model.LinearRegression()
```

```python
#reshape error
#predicted = lr.fit(X=tips['total_bill'], y =tips['tip'])
```

```python
predicted = lr.fit(X=tips['total_bill'].values.reshape(-1,1), y =tips['tip'])
predicted.intercept_
```

```python
predicted.coef_
```

```python
pred2 = lr.fit(X=tips[['total_bill', 'size']], y = tips['tip'])
pred2.coef_
```

```python
pred2.intercept_
```

```python
tips_dummy = pd.get_dummies(tips[['total_bill', 'size', 'smoker', 'day', 'time']])
tips_dummy
```

```python
tips_dummy_ref = pd.get_dummies(tips[['total_bill', 'size', 'smoker', 'day', 'time']], drop_first=True)
tips_dummy_ref.head()
```

```python
lr2 = linear_model.LinearRegression()
pred3 = lr2.fit(X=tips_dummy_ref, y = tips['tip'])
pred3.coef_
```

```python
values = np.append(pred3.intercept_, pred3.coef_)
names = np.append('intercept', tips_dummy_ref.columns)
```

```python
results4 = pd.DataFrame(values, index=names, columns = ['coef'])
results4
```
