```python
def main():
    value = 99
    change_me(value)
    print(f"value = {value} in main")

def change_me(value):  
    value = 0 
    print(f"value = {value} in change_me")
        
if __name__ == '__main__':
    main()

# value is passed by value, so the change in change_me does not affect the value in main
# int, float, bool, str, tuple are immutable types, so they are passed by value
```

```python
def main():
    value = 99
    value = change_me(value) # value changed here
    print(f"value = {value} in main")

def change_me(value):  
    value = 0 
    print(f"value = {value} in change_me")
    return value
        
if __name__ == '__main__':
    main()
```

```python
def is_even(number: int) -> bool:
    return number % 2 == 0
```

```python
def test_is_even():
    print("Testing is_even(27)")
    expected = False
    actual = is_even(27)
    if expected == actual:
        print("Test passed")
    else:
        print("Test failed")
        print("Expected: ", expected)
        print("Actual: ", actual)
        
test_is_even()
```

```python
age = 19
def main():
    vote_determiner()
    print("Age:", age)
def vote_determiner():
    global age # this changes the global variable
    age = int(input("Enter your current age: "))
    if age >= 18:
        print("You are eligible to vote!")
    else:
        print("You are not eligible to vote, sorry!")

if __name__ == '__main__':
    main()
```
