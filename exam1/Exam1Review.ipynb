```python
# tor print with commas
num = 400000000
print(f"{num:,}") # add commas
print(f"{num:.2f}") # 2 decimal places
# types: f, d, e, s
# f: floating point
# d: decimal (integer)
# e: exponential
# s: string
print(f"{num:d}") # integer
dec = 123456789.123456789 
print(f"{dec:.2f}") # 2 decimal places
print(f"{dec:.2e}") # exponential

word = "hello"
print(f"{word:.2s}") # string, 2 characters
print(f"{word:>10}") # right aligned (pushes to right because it points to right), width 10, 5 spaces in front
print(f"{word:<10} dog") # left aligned, width 10. then 'dog' is added after the width of 10
print(f"{word:^10}") # center aligned, width 10
print(f"{word:*>10}") # right aligned, width 10, * padding
print(f"{word:*<10}") # left aligned, width 10, * padding
print(f"{word:*^10}") # center aligned, width 10, * padding
```

```python
def add_up():
    sum = 0
    try:
        while True:
            num = int(input("Enter an int. Positive to sum, -1 to end: "))
            if num < -1:
                raise ValueError
            elif num == -1:
                break
            else:
                sum = sum + num
    except:
        print("Enter a valid input")

    print(sum)

add_up()
```

```python
def char_dict(s:str) -> dict:
    d = dict() # or {}
    for x in range(0, len(s)):
        c = s[x]
        val = d.get(c, 0) # or d[c] = 1 ...
        # d.update(c, val + 1) used wrong. Supposed to do d.update({c: val + 1})
        # otherwise do 
        d[c] = val + 1
    return d

char_dict("Snow, snow, go away!")
```

```python
password = "PASSwd"

newpass = ""

has_alpha = False

if len(password) < 3:
    newpass = "Ab12" + password
else:
    

    for c in password: 
        if c.isupper():
            newpass += c + c.lower()
            has_alpha = True
        elif c.islower():
            newpass += c + c.upper()
            has_alpha = True
        else:
            newpass += c

if not has_alpha:
    newpass = "Ab12" + password

print(newpass + str(len(password)))

```
