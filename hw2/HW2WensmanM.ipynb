```python
"""
 Author: Malik Wensman
 Email: mdwensma@ncsu.edu
 Class: CSC 251-001 Fall 2026
 Program: Yardage Calculator
 Purpose: Compute the yardage needed for a given amount of money.
 Bugs: None
 Acknowledgements: None
"""

import math
from fractions import Fraction

# CONSTANTS
WIDTH = 21.5 # inches
YARD_INCHES = 36
ROUNDING = 8 # round to the 1/8 of a yard. You always roundup

# user selects length of fabric in yards 
# to get the length of fabric you need to cut, you add the 0.5 in seam allowance on each of the two ends (top and bottom)
# must calculate the finished_length (what is left after sewing)
finished_length_in_inches = float(input('Enter the length of the pillowcase in inches (0-108"): '))
while finished_length_in_inches <= 0 or finished_length_in_inches > 108:
    finished_length_in_inches = float(input('Invalid input for length. Re-enter the length in inches(0-108"): '))
```

```python
finished_length_in_inches = finished_length_in_inches + 1 # add 0.5 inch seam allowance to each end (top and bottom)
finished_length_in_yards = finished_length_in_inches / YARD_INCHES 
print(finished_length_in_yards)
finished_length_in_yards = math.ceil(finished_length_in_yards * ROUNDING) / ROUNDING # round up to the nearest 1/8 of a yard
finished_length_in_yards
```

```python
# calculate the price

price_per_yard = float(input('Enter the price per yard: '))
while price_per_yard < 0:
    price_per_yard = float(input('Invalid input for price. Re-enter the price per yard: '))

cost = finished_length_in_yards * price_per_yard
```

```python
finished_length_whole_yards = int(finished_length_in_yards) # get the whole yards
print(finished_length_whole_yards)

finished_length_leftover_yards = finished_length_in_yards - finished_length_whole_yards # get the leftover yards
print(finished_length_leftover_yards)

finished_length_leftover_eighths = round(finished_length_leftover_yards * ROUNDING) # get the leftover yards and convert to eighths
print(finished_length_leftover_eighths)



fraction_obj = Fraction(finished_length_leftover_eighths, ROUNDING) # create a fraction object
print(f"Minimum fabric needed: {str(finished_length_whole_yards) + "-" if finished_length_whole_yards > 0 else ''}", end="")
print(f"{fraction_obj.numerator}/{fraction_obj.denominator} yards" if finished_length_leftover_eighths > 0 else "")
    
```

```python
print(f"Fabric cost: ${cost:.2f}")
```
