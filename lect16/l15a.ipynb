Load the GlobalLandTemperatures.csv file into a DataFrame in your notebook. How many observations are included in the file?

```python
import pandas as pd
df = pd.read_csv('GlobalTempData.csv')
df
```

How many missing values (NaN) occur in the dataset?

```python
Nan_count = df.isnull().sum()
print(Nan_count)
print(Nan_count.sum())
```

Filter the data by the city specified. How many observations are left for that city once the NaNs are removed?

```python
df = df.dropna()
df
```

Compute the average temperature, in degrees Celsius, for the city chosen in class. Include 1 decimal value in your answer.

```python
avg_temp = df['AverageTemperature'].mean()
print(f"The average temperature is: {avg_temp:.1f} °C")
```

```python
jakarta_rows = df[df['City'] == 'Jakarta']
jakarta_rows
```

```python
jakarta_avg_temp = jakarta_rows['AverageTemperature'].mean()
print(f"The average temperature in Jakarta is: {jakarta_avg_temp:.1f} °C")
```
