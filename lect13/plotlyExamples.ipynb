```python
import pandas as pd
import plotly.express as px 
tips = pd.read_csv('tips.csv')
tips.head()
```

```python
tdf = tips.groupby('size')['tip'].mean
tdf
```

```python
fig = px.bar(tips, x="size", y="tip") 
fig.show()
```

```python
tips["percent"] = tips["tip"]/tips["total_bill"]*100
tips
```

```python
newtip = tips[["size", "percent"]]
newtip
```

```python
# Creating the Figure instance
fig = px.scatter(newtip, x="size", y="percent") 
 
# showing the plot
fig.show()
```

```python
df = px.data.tips()
df
```

```python
# Create the figure:
# x-axis: 'sex', y-axis: 'total_bill'
# color the bars by 'smoker' status, group them by 'day'
fig = px.bar(df, x="sex", y="total_bill", color="smoker", barmode="group", facet_col="day")

# Display the interactive figure
fig.show()
```
