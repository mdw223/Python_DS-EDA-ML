```python
# You wish to strip off the first 2 characters of each string in a list of strings, 
# keeping the remaining part of the string. Create a map to perform this operation. 
# (Use slist as your list of strings and first_two to hold the iterable)

def first_two(s):
    return s[2:]

slist = ["Python", "C", "Java", "C++"]

mapper = map(first_two, slist)
print(list(mapper))
```
