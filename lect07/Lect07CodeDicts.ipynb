Lecture 07 Practice

Write the code to create a dictionary with a name as the key and a list containing a GPA and major as the value. For example:

pythonclass ={'Ann': [3.4, 'CSC'], 'Bob': [2.8, 'MTH'], 'Jim': [3.9, 'ECE']}

Consider how the names, GPAs,  and majors will be stored. You may assume an equal number of each. Your code should work for any number of key-value pairs.

```python
name = ['Ann', 'Bob', 'Jim']
GPA = [3.4, 2.8, 3.9]
major = ['CSC', 'MTH', 'ECE']
values = zip(GPA, major)
dictComp = {k:list(v) for (k,v) in zip(name, values)}
print(dictComp)
```

```python
name = ["Ann", "Bob", "Jim"]
GPA = [3.4, 2.8, 3.9]
major = ["CSC", "MTH", "ECE"]
d = {}
for i in range(len(name)):
    d[name[i]] = [GPA[i], major[i]]
d
```
