## Lecture 15 Notebook - Combining Datasets

```python
import pandas as pd
```

```python
df1 = pd.read_csv('df1.csv')
df2 = pd.read_csv('df2.csv')
df1
```

```python
df2
```

```python
row_concat = pd.concat([df1, df2])
row_concat
```

```python
row_concat = pd.concat([df1, df2], ignore_index=True)
row_concat
```

```python
col_concat = pd.concat([df1, df2], axis=1)
col_concat
```

```python
col_concat['newcol'] = ['a', 'b', 'c', 'd']
col_concat
```

```python
col_concat = pd.concat([df1, df2], axis=1, ignore_index=True)
col_concat
```

```python
df_merge = df1.merge(df2)
df_merge
```

```python
df = pd.read_csv('petdata.csv')
df
```

```python
df = df.drop(columns=['Unnamed: 0'], axis=1)
df
```

```python
breeders = pd.read_csv('pet_breeder.csv')
breeders
```

```python
petbreeder_merge = df.merge(breeders)
petbreeder_merge
```

```python
owners = pd.read_csv('pet_owner.csv')
owners
```

```python
badmerge = df.merge(owners, how='outer')
```

```python
pet_merge = df.merge(owners, left_on='petname', right_on = 'Dogname')
pet_merge
```

```python
pet_merge_left = df.merge(owners, left_on='petname', right_on = 'Dogname', how = 'left')
pet_merge_left
```

```python
pet_merge_left2 = df.merge(breeders, how = 'left')
pet_merge_left2
```

```python
pet_merge_right = df.merge(owners, left_on='petname', right_on = 'Dogname', how = 'outer')
pet_merge_right
```

```python
pet_merge_right2 = df.merge(breeders, how = 'outer')
pet_merge_right2
```
