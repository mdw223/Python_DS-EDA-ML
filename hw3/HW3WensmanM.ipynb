```python
"""
 Author: Malik Wensman
 Email: mdwensma@ncsu.edu
 Class: CSC 251-001 Fall 2026
 Program: Conversion table
 Purpose: Produce a table that converts heights in feet and inches to centimeters
 Bugs: None
 Acknowledgements: None
"""


#prompt user to enter the range of values they wihs for the table to span
# first a starting valye for the height in feet and inches
# then the ending value in feet and inches (assuming the table spans form a shorter height to a larger height)

# give your user info concerning the rnage of values to be entered and allow them to enter either an integer or a flaot
# round values to nearest integer 

# output a well aligned formatted table to span from the beiginning height up to and including the end height,
# incrementing 1 inch per line
# the table header is provied in header.py (import this as header())
# the height in feet and inches in the first two columns should be output with no floating-point values
# the centimenter conversion should output with 2 decimal places
# all numberic values should be place value aligned.

# Requirements:
# a file named hw3functions.py that includes at least the following functions:
# get_values() - prompts the user for a height in feet and inches and returns 2 acceptable values
# convert_values() that is passed 2 values, feet and inches, returns the height in centimeters, assuming 2.54cm per inch
# print_table() that prints out a single formatted row of values for the table

# hw3main.py has main() and should import header and hw3functions modules and call their functions as needed 

# input validation to test the user's values are within range and that they don't enter ending values that are smaller than the starting values.
# allow the user to enter both the feet and inches before reprompting them for both values if one of the values is not valid
# assume reasonable values range from 0 feet, 0 inches up to values too large to be fit in the table format
# feet and inches should be entered such that the number of inches does not exceed 11 (i.e. 5 feet 15 inches should be 6 feet 3 inches)
# if they enter 12 inches or more, prompt them to reenter BOTH values
# if the ending values they enter are smaller than the starting values, prompt them to enter both ending values agian
# no exception handling is required
# assume user will enter numberic values 
# all floats should be rounded to to integers.

# Document your files 
# each file with an opening comment , each function

# global constants that replace magic numbers

# zip togther header.py , hw3functions.py, and hw3main.py to produce a complete program that meets the requirements of the assignment



```

```python
INCHES_PER_FOOT = 12
INCHES_TO_CM = 2.54

'''
Type can be 'starting' or 'ending' to indicate which height is being entered
'''
def get_values(type: str, start_feet: int, start_inches: int) -> tuple[int, int]: 
    feet = input(f"Enter the {type} height in feet and inches. First enter the number of feet as an integer between 0 and 20: ")
    feet = int(feet)

    inches = input("Now enter the number of inches as an integer between 0 and 11: ")
    inches = int(inches)

    while feet < 0 or feet > 20 or inches < 0 or inches > 11:
        print("Invalid input. Please enter a value between 0 and 20 for feet and a value between 0 and 11 for inches.")
        feet = input(f"Enter the {type} height in feet and inches. First enter the number of feet as an integer between 0 and 20: ")
        feet = int(feet)
        inches = input("Now enter the number of inches as an integer between 0 and 11: ")
        inches = int(inches)

    if type == 'ending':
        if feet < start_feet or (feet == start_feet and inches < start_inches):
            print("Invalid input. The ending height must be greater than or equal to the starting height.")
            return get_values(type, start_feet, start_inches)

    return feet, inches

def convert_values(feet: int, inches: int) -> float: 
    total_inches = feet * INCHES_PER_FOOT + inches
    centimeters = total_inches * INCHES_TO_CM
    return centimeters


def print_table(feet: int, inches: int, centimeters: float) -> None:
    print(f"{feet:>10} | {inches:>10} || {centimeters:>10.2f}")

def header():
  """Provides formatting for table header
      Parameters: none
      Returns: none
  """
  print(f"{'Height Conversion Table':^40}")
  print(f"{'feet':^10} | {'inches':^10} || {'cm':^10}")
  print("=====================================")
```

```python
header()

start_feet, start_inches = get_values('starting', 0, 0)
end_feet, end_inches = get_values('ending', start_feet, start_inches)

current_feet = start_feet
current_inches = start_inches

while(current_feet < end_feet or (current_feet == end_feet and current_inches <= end_inches)):
    centimeters = convert_values(current_feet, current_inches)
    print_table(current_feet, current_inches, centimeters)

    current_inches += 1
    if current_inches == INCHES_PER_FOOT:
        current_inches = 0
        current_feet += 1
```
