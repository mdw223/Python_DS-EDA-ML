The orange tree data provided describes the growth of several trees planted on 12/31/1968. Their ages are given in terms of days since they were planted and their circumference is given in mm, measured at chest height. 

Find the correlation (r-value) between age and circumference. Enter your value to 4 decimal places as a decimal, not a percent.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv('Orange.csv')
data.head()

import seaborn as sns 
sns.regplot(x='age', y = 'circumference', ci = None, data = data) 
# r value
print('r value:', data['age'].corr(data['circumference']))
```

Enter the equation of the regression line for the trees in the Orange dataset, where age is the independent variable and circumference is the dependent variable. Include all decimal values given.

```python
# equation of the regression line
X = data['age']
Y = data['circumference']
# slope (m) and intercept (b)
m, b = np.polyfit(X, Y, 1)
print('slope:', m)
print('intercept:', b)
print(f'equation of the regression line: y = {m}x + {b}')
```
