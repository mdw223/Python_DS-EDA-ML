```python
def count_T1(my_string):
    count = 0
    for ch in my_string:
        if ch == 'T' or ch  == 't':
            count += 1
    return count

print(count_T1("Titus"))
```

```python
def count_T2(my_string):
    index = 0	
    count = 0						
    while index < len(my_string):
        if my_string[index] == 'T' or my_string[index] == 't':
            count += 1
        index += 1
    return count
print(count_T2("Titus"))
```

```python
def count_T3(my_string):
    return my_string.upper().count('T')

print(count_T3("Titus"))
```

```python
# What value is returned by the valid_password function in Example 12 for the following input?

def valid_password(password):
    # Set Boolean flags to false
    correct_length = False
    has_uppercase = False
    has_lowercase = False
    has_digit = False
    # Test the password's length
    if len(password) >= 7:
        correct_length = True
    # Test each character and adjust flag value
    for ch in password:
        if ch.isupper():
            has_uppercase = True
        if ch.islower():
            has_lowercase = True
        if ch.isdigit():
            has_digit = True
    # Return true if all requirements are met; false if not
    return correct_length and has_uppercase and has_lowercase and has_digit

valid_password("Wolf25")
```
