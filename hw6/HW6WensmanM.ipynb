Import numpy and matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

Create two identical 1D ndarrays, x and y, to hold grid values from -.5 m up through .5 (inclusive) 
with an interval of 0.01 m

```python
# two identical 1D ndarrays
x = np.arange(-0.5, 0.51, 0.01)
y = np.arange(-0.5, 0.51, 0.01)
```

Create a NumPy meshgrid to create a grid of points using the x and y arrays

```python
# numpy meshgrid, has a grid of porints using the x and y arrays
xx, yy = np.meshgrid(x, y)
```

Position particle 1 at (0.105, 0.105) with a charge of 1 nanocoulombs. Offsetting the particle from a grid point will prevent any divide by zero errors.

```python
# position particle 1 at this point with its charge in nanocoulombs
p1 = (0.105, 0.105)
p1_c = 1e-9 # in nanocoulombs
```

Compute the electric potential due to particle 1 for each point of the meshgrid

```python
# Compute the electric potential due to particle 1 for each point of the meshgrid.
k = 9e9
# distance from particle 1 to each point in the meshgrid
r = np.sqrt((xx - p1[0])**2 + (yy - p1[1])**2) 

zz = p1_c * k / r
```

Use a contour map to illustrate the electric potential values around charge 1.

```python
plt.contourf(x,y,zz) # contour plot of the electric potential
plt.axis('scaled')
plt.colorbar()
plt.title("Electric Potential Due to 1nC Charge")

# make y axis label as 'y-axis (m)'
plt.ylabel('y-axis (m)')

# make x axis label as 'x-axis (m)'
plt.xlabel('x-axis (m)')
```

Position particle 2 at (0.305, -0.305) with a charge of 2 nanocoulombs. 

```python
p2 = (0.305, -0.305)
p2_c = 2e-9 # in nanocoulombs
```

Compute the electric potential due to particle 2 for each point of the meshgrid

```python
# electric potential due to particle 2
r2 = np.sqrt((xx - p2[0])**2 + (yy - p2[1])**2)
zz2 = p2_c * k / r2
```

create a contour map of the electric potential values. 

```python
plt.contourf(x,y,zz2) # contour plot of the electric potential
plt.axis('scaled')
plt.colorbar()
plt.title("Electric Potential Due to 2nC Charge")

# make y axis label as 'y-axis (m)'
plt.ylabel('y-axis (m)')

# make x axis label as 'x-axis (m)'
plt.xlabel('x-axis (m)')
```

Compute the total electric potential for all points in the region.

```python
# Compute the total electric potential for all points in the region.

zz_total = zz + zz2
```

Use a contour map to show both charges and the point (0,0) as well as the electric potential values 
throughout the meshgrid. 

```python
# Use a contour map to show both charges and the point (0,0) 
# as well as the electric potential values throughout the meshgrid

plt.contourf(x,y,zz_total) # contour plot of the total electric potential
plt.axis('scaled')
plt.colorbar()
plt.title("Electric Potential Due to 1nC and 2nC Charges")

# make y axis label as 'y-axis (m)'
plt.ylabel('y-axis (m)')

# make x axis label as 'x-axis (m)'
plt.xlabel('x-axis (m)')

plt.plot(0, 0, 'go', markersize=10)  # plot the point (0,0)
```

Output the total electric potential at the point (0, 0) using code to output the value rather than visually picking the value out of the array

```python
# testing to get the total electric potential using the array

total_potential_at_origin = zz_total[50, 50]  # index (50, 50) corresponds to the point (0, 0)
print(f"Total electric potential at the point (0, 0): {total_potential_at_origin} V")
```

```python
# Output the total electric potential at the point (0, 0) using code to output the value 
# rather than visually picking the value out of the array

# to get the total electric potential at the point (0, 0) is to use the formula 
# for electric potential due to multiple point charges:
total_potential_at_origin_formula = (p1_c * k / np.sqrt((0 - p1[0])**2 + (0 - p1[1])**2)) + (p2_c * k / np.sqrt((0 - p2[0])**2 + (0 - p2[1])**2))
print(f"Total electric potential at the point (0, 0) using formula: {total_potential_at_origin_formula} V")
```
