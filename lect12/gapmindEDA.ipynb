```python
# Filter for Europe
europe = df[df['continent'] == 'Europe']

# Mean gdpPercap per year across all European countries
europe_gdp = europe.groupby('year')['gdpPercap'].mean()

print(europe_gdp)
```

```python
change = europe_gdp.iloc[-1] - europe_gdp.iloc[0]
print(f"Change in avg GDP per cap (1952–2007): {change:.2f}")
```
