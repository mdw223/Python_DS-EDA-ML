```python
import csv
```

```python
books = [["The Nightingale", 2015, "Kristin Hannah"], 
        ["Teaching Students to Learn", 2015, "Saundra McGuire"],
        ["Hidden Figures", 2016, "Margot Lee Shetterly"]]


with open('booklist.csv', 'w', newline = '') as outfile:
	writer = csv.writer(outfile)
	writer.writerows(books)
```

```python
with open('booklist.csv', 'r', newline = '') as infile:
	reader = csv.reader(infile)
	for row in reader:
	    print(row)
```

```python
with open('book_table.csv', 'r', newline = '') as my_file:
    my_data = csv.DictReader(my_file)
    #print(my_data)
    for line in my_data:
        print(line)
        #print(line['Title']+ ' by '+ line['Author'], line["Year Published"])
```

```python
with open('book_csv.csv', 'w') as this_file:
    cols = ['Title', 'Year Published', 'Author']
    writer = csv.DictWriter(this_file, fieldnames=cols)
    writer.writeheader()
    writer.writerow({'Title': 'The Nightingale', 'Year Published': '2015', 
			'Author': 'Kristin Hannah'})
    writer.writerow({'Title': 'Teaching Students to Learn', 
			'Year Published': '2015', 'Author': 'Saundra McGuire'})
    writer.writerow({'Title': 'Hidden Figures', 'Year Published': '2016', 
			'Author': 'Margot Lee Shetterly'})
```

```python
import pickle
with open('book.bin', 'wb') as fout:
	 pickle.dump(books, fout)
```

```python
with open('book.bin', 'rb') as fin:
	book_list = pickle.load(fin)
	print(book_list)
```
