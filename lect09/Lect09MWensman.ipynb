```python
"""
 Author: Malik Wensman
 Email: mdwensma@ncsu.edu
 Class: CSC 251-001 Fall 2026
 Program: Owid Energy Data Analysis
 Purpose: create a dictionary , key as country name, value tuple for percent change in energy consimption and greenhouse gas emissions as measured in megtonnes of co2 equivalents
 Bugs: None
 Acknowledgements: None
"""

from typing import Any


import csv

with open('owid-energy-data.csv') as my_file:
    my_data = csv.DictReader(my_file)
    dict = {}
    for line in my_data:
        if line['country']:
            dict[line['country']] = (float(line['energy_cons_change_pct']), float(line['greenhouse_gas_emissions']))
    print(dict)
```
