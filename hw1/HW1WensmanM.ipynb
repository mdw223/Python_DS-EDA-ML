## Instructions

Write a program to prompt the user for the total amount of change that they have in dollars and cents, entered as a
single floating-point value, then compute the least number of coins that they could possibly have if their change
contains only dollar coins, half-dollars, quarters, dimes, nickels, and pennies. Output the total number of coins as well as
how many of each type of coin in a properly aligned 2-column table as shown below.

## Example

Your program should match the following when executed. User input is shown in bold. Each line of equal signs or
hyphens is 26 characters long.

```bash
Enter the total amount of money: $12.83
Fewest coins for $12.83
==========================
Coin Type Number
--------------------------
Dollars 12
Half-Dollars 1
Quarters 1
Dimes 0
Nickels 1
Pennies 3
--------------------------
Total coins 18
```

For the identical input of $12.83, the following output would be INCORRECT (2 dimes and 1 nickel makes a quarter):

```bash
Enter the total amount of money: $12.83
Fewest coins for $12.83
==========================
Coin Type Number
--------------------------
Dollars 12
Half-Dollars 1
Quarters 0
Dimes 2
Nickels 2
Pennies 3
--------------------------
Total coins 20
```

Three sample input and expected output files are provided for you. You should create several more tests to make sure
that your code always provides the correct answer. You’ll notice that the user input is not present in the files as in the
example above since the files were created with I/O redirection. Your spacing and user prompt do not need to match
exactly but should be clear and should include appropriate units ($). Your table, however, should be formatted
identically as the one provided and your numeric values in the table should remain aligned.
